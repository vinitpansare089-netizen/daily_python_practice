```python
from numbers import Number

def filter_models(scores: dict, threshold: float) -> dict:

    if not isinstance(scores, dict):
        raise TypeError("Input must be dict")

    if len(scores) == 0:
        raise ValueError("Dict must not be empty")

    if not isinstance(threshold, Number):
        raise TypeError("Threshold must be a number")

    result = {}

    for key, value in scores.items():

        if not isinstance(key, str):
            raise TypeError("Keys must be strings")

        if not isinstance(value, Number):
            raise TypeError("Values must be numbers")

        if value >= threshold:
            result[key] = value

    return result

scores = {"A":0.8, "B":0.9, "C":0.85}

print(filter_models(scores, 0.9))
```
