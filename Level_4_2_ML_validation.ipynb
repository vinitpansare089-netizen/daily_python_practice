```python
from numbers import Number

def scale_scores(scores: dict) -> dict:
  ###Validation first VINIT...
  if not isinstance(scores, dict):
    raise TypeError("Input must be Dict")

  if len(scores) == 0:
    raise ValueError("Input should not be empty")

  for key, value in scores.items():
    if not isinstance(key, str):
      raise TypeError("Keys must be string")

    if not isinstance(value, Number):
      raise TypeError("Values must be Number")
####finding min and max values
  values = list(scores.values())

  min_value = min(values)
  max_value = max(values)

  if min_value == max_value:
      raise ValueError("min and max value should not be same")

####stores value in result
  result = {}

  for key, value in scores.items():
    new_value = (value - min_value) / (max_value - min_value)
    result[key] = new_value

  return result

scores = {
    "A": 40,
    "B": 20,
    "C": 10,
    "D": 80
}

print(scale_scores(scores))
# {'A': 0.42857142857142855, 'B': 0.14285714285714285, 'C': 0.0, 'D': 1.0}
```
