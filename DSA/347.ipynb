```python
"""
Top k frequent elements
"""

nums = [1,1,1,2,2,3]
k = 1

count = {}

for num in nums:
    count[num] = count.get(num, 0) + 1

result = sorted(count, key=count.get, reverse=True)

print(result[:k])
```

```python
nums = [1,1,1,2,2,3]
k = 2

count = {}

for num in nums:
    if num in count:
        count[num] += 1
    else:
        count[num] = 1


result = sorted(count, key=count.get, reverse=True)

print(result[:k])
```
