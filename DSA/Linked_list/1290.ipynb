```python
"""
101 -> 5 
Binary to decimal number using linked list
"""
class listNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

#############################################
head = [1,0,1]

current = head

while current:
    # number = number * 2 + current.val
    current = current.next

print(current)
```
