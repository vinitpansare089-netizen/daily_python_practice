```python
class Solution:
    def reorderList(self, head: Optional[ListNode]) -> None:

        if not head or not head.next:
            return

        # 1. Find middle
        slow = head
        fast = head

        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

        # 2. Split list
        second = slow.next
        slow.next = None

        # 3. Reverse second half
        prev = None
        current = second

        while current:
            temp = current.next
            current.next = prev
            prev = current
            current = temp

        # 4. Merge
        first = head
        
        second = prev

        while second:
            temp1 = first.next
            temp2 = second.next

            first.next = second
            second.next = temp1

            first = temp1
            second = temp2
```
