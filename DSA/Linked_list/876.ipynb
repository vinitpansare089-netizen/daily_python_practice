```python
"""
find the middle number in linked list
"""
class listNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class solution:
    def middldenode(self, head: listNode) -> listNode:

        slow = head
        fast = head 

        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

        return slow
        # print(slow)

```
