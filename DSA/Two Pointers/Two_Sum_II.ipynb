```python
"""
Two Sum II
"""

class solution:
    def twoSum(self, numbers:list[int], target:int) -> list[int]:
    
        left = 0
        right = len(numbers) - 1

        while left < right:

            sum = numbers[left] + numbers[right]

            if sum == target:
                return [left + 1, right + 1]
            
            elif sum < target:
                left += 1
    
            else:
                right -= 1

obj = solution()

result = obj.twoSum([2, 11, 7, 15], 9)
print(result)
```
