```python
"""
word pattern
"""
pattern = "abba" 
s = "dog cat cat dog"

words = s.split()

if len(pattern) != len(words):
    print(False)

seen = {}
seen1 = {}

for i in range(len(pattern)):
    if pattern[i] in seen:
        if seen[pattern[i]] != words[i]:
            print(False)


        if words[i] in seen1:
            if seen1[words[i]] != pattern[i]:
                print(False)

```
