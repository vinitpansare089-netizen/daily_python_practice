```python
##Finding maximum element

arr = [12, 45, 7, 89, 23, 56]

max = arr[0]

for num in arr:
    if max < num:
        max = num
print(max)
```

```python
##Finding Second Largest element

arr = [90, 45, 7, 1, 23, 56]

if arr[0] > arr[1]:
    largest = arr[0]
    second_largest = arr[1]
else:
    largest = arr[1]
    second_largest = arr[0]

for num in arr[2:]:
    if num > largest:
        second_largest = largest
        largest = num

    elif num > second_largest:
        second_largest = num

print(largest)
print(second_largest)
```

```python
#Finding Minimum Element 

arr = [90, 45, 7, 5, 23, 1]

minimum = arr[0]
min_index = 0

for i, num in enumerate(arr):
    if num < minimum:
        minimum = num
        min_index = i
    
print(f'Number: {minimum} -> Index: {min_index}')
```
