```python
##Finding maximum element

arr = [12, 45, 7, 89, 23, 56]

max = arr[0]

for num in arr:
    if max < num:
        max = num
print(max)
```

```python
##Finding Second Largest element

arr = [90, 45, 7, 1, 23, 56]

if arr[0] > arr[1]:
    largest = arr[0]
    second_largest = arr[1]
else:
    largest = arr[1]
    second_largest = arr[0]

for num in arr[2:]:
    if num > largest:
        second_largest = largest
        largest = num

    elif num > second_largest:
        second_largest = num

print(largest)
print(second_largest)
```

```python
#Finding Minimum Element 

arr = [90, 45, 7, 5, 23, 1]

minimum = arr[0]
min_index = 0

for i, num in enumerate(arr):
    if num < minimum:
        minimum = num
        min_index = i
    
print(f'Number: {minimum} -> Index: {min_index}')
```

```python
##Even Numbers count
arr = [90, 45, 7, 5, 23, 1, 8, 10]

count = 0

for num in arr:
    if num % 2 == 0:
        count += 1

print(count)
```

```python
###Odd number count

arr = [90, 45, 7, 5, 23, 1, 8, 10]

even = 0
odd = 0

for num in arr:
    if num % 2 == 0:
        even += 1
    else:
        odd += 1

print(f'Even: {even}')
print(f'Odd: {odd}')
```

```python
##Count positive and negetive

arr = [-2, 5, -8, 9, 0, 3, -1]

positive = 0
negetive = 0

for num in arr:
    if num > 0:
        positive += 1
    elif num < 0:
        negetive += 1

print(positive)
print(negetive)
```
