```python
###Brute force

def hashmapping(nums, target):

    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i, j]
            return False
            
nums = [6,5,9,1]
target = 12

print(hashmapping(nums, target))
```

```python
###Hashmapping

def hashmapping(nums, target):
    seen = {}
    for i, num in enumerate(nums):
        needed = target - num
        # print(seen)
        if needed in seen:
            return [seen[needed], i]
        else:
            return 'No Key Found'


nums = [12,4,2,5]
target = 9

print(hashmapping(nums, target))
```
