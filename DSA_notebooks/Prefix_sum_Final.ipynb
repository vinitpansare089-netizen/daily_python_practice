```python
"""
Prefix sum
"""

arr = [2, 4, 1, 6, 3]

total = 0

for num in arr:
    total = total + num

print(total)
```

```python
"""
Running Total prefix sum
"""
arr = [2,4,1,6,3]

total = 0
prefix = []

for num in arr:
    total += num
    prefix.append(total)


print(prefix)

L = 0
R = 3

if L == 0:
    ans = prefix[R]
else:
    ans = prefix[R] - prefix[L - 1]

print(ans)
```

```python
"""
Pivot Index
"""

nums = [1,1,3,6,5,6]

total_sum = sum(nums)
left_sum = 0

for i, num in enumerate(nums):

    right_sum = total_sum - left_sum - num

    if left_sum == right_sum:
        print(i)
        break

    left_sum += num
    

else:
    print(-1)
```
