```python
stack = []

pairs = {
    ')' : '(',
    ']' : '[',
    '}' : '{'
}

def stacky(s: str):
    for char in s:
        
        if char in '([{':
            stack.append(char)

        else:
            if not stack:
                return False
        
            if stack[-1] != pairs[char]:
                return False
            
            stack.pop()
        
        
    return len(stack) == 0

s = "()[]{}"

print(stacky(s))
```
