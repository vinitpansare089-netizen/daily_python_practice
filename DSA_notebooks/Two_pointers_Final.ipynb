```python
##Two pointers 
##Reverse an Array


arr = [10,20,30,40,50]

left = 0
right = len(arr) - 1

while left < right:
    temp = arr[left]   ###Created using temporary variable
    arr[left] = arr[right]
    arr[right] = temp
    left += 1
    right -= 1

print(arr)


###Second method 

arr = [10,20,30,40,50]

left = 0
right = len(arr) - 1

while left < right:
    arr[left], arr[right] = arr[right], arr[left]
    left += 1
    right -= 1

print(arr)
```

```python
###Reverse String

s = ["h","e","l","l","o"]

left = 0
right = len(s) - 1

while left < right:
    s[left], s[right] = s[right], s[left]
    left += 1
    right -= 1

print(s)
```

```python
### Palindrome 
"""
we are supping this is a palindrome
"""

x = 124

s = str(x)

left = 0
right = len(s) - 1

is_palindrome = False

while left < right:
    if s[left] == s[right]:
        is_palindrome = True
        break

    left += 1
    right -= 1

print(is_palindrome)
```

```python
### Valid Palindrome

s = "A man, a plan, a canal: Panama"

left = 0
right = len(s) - 1

while left < right:

    if not s[left].isalnum():
        left += 1

    elif not s[right].isalnum():
        right -= 1

    elif s[left].lower() != s[right].lower():
        print(False)
        break

    else:
        left += 1
        right -= 1

else:
    print(True)
```
