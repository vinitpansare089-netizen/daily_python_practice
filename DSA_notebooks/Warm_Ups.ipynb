```python
nums = [4, 7, 2, 9, 5]

max = nums[0]

for num in nums:
    if num > max:
        max = num

print(max)
```

```python
nums = [1, 2, 3, 4, 5, 6]

left = 0
right = len(nums) - 1

while left < right:
    nums[left], nums[right] = nums[right], nums[left]
    left += 1
    right -= 1

print(nums)
```

```python
nums = [1, 2, 2, 3, 2, 4, 1]

freq = {}

for num in nums:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1

print(freq)

# for num in freq:
#     if freq[num] >= len(nums) // 2:
#         break

for num,count in freq.items():
    if count >= len(nums) // 2:
        print(num, count)

# print(freq[num])
```

```python
nums = [2, 5, 8, 12, 16, 21, 30]
target = 22

left = 0
right = len(nums) - 1

while left <= right:
    mid = (left + right) // 2

    if nums[mid] == target:
        print(True)
        break
    
    elif nums[mid] < target:
        left = mid + 1

    elif nums[mid] > target:
        right = mid - 1
else:
    print(False)
```
