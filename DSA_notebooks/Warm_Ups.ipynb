```python
nums = [4, 7, 2, 9, 5]

max = nums[0]

for num in nums:
    if num > max:
        max = num

print(max)
```

```python
nums = [10, 20, 5, 8, 2, 9]

min = nums[0]

for num in nums:
    if num < min:
        min = num

print(min)
```

```python
nums = [1, 2, 3, 4, 5, 6]

left = 0
right = len(nums) - 1

while left < right:
    nums[left], nums[right] = nums[right], nums[left]
    left += 1
    right -= 1

print(nums)
```
