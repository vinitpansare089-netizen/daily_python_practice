```python
"""
Priority queue
"""

import heapq

pq = []

heapq.heappush(pq,-5)
heapq.heappush(pq,-2)
heapq.heappush(pq,-8)
heapq.heappush(pq,-1)

while pq:
    print(-heapq.heappop(pq))
```

```python
"""
By providing hospital situation where patient comes with a priority.
"""
N = int(input())
arr = list(map(int, input().split()))

import heapq

arr = [-x for x in arr]
heapq.heapify(arr)

while arr:
    print(-heapq.heappop(arr))
```

```python
"""
Maximum priority queue
"""

N = int(input())
arr = list(map(int, input().split()))

import heapq

pq = []

for num in arr:
    #heapq.heappush(pq, -num)
    heapq.heappush(pq, num)

while pq:
    print(heapq.heappop(pq))

```
