```python
"""
Find the most frequent number
"""

nums = [4, 2, 4, 5, 2, 4, 3]

freq = {}

for num in nums:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1

print(freq)

count = 0
frequent_num = None

for num in freq:
    if freq[num] > count:
        count = freq[num]
        frequent_num = num

print(frequent_num, count)
```

```python
nums = [3, 7, 11, 15, 19, 24, 30]
target = 3

left = 0
right = len(nums) - 1

while left <= right:

    mid = (left + right) // 2

    if nums[mid] == target:
        print(True)
        break

    elif nums[mid] < target:
        left = mid + 1

    else:
        right = mid - 1

else:
    print(False)
 
```
