```python
arr = [45, 32, 31, 24, 65, 77]

for num in arr:
    if num > 1:
        is_prime = True

        for i in range(2, num):
            if num % i == 0:
                is_prime = False
                break

        if is_prime:
            print(num)
```

```python
arr = [22, 11, 34, 45, 55, 76, 67, 27]

arr1 = []
arr2 = []

for num in arr:
    if num % 2 == 0:
        arr1.append(num)

    else:
        arr2.append(num)

arr3 = arr1 + arr2
print(arr3) 
```
