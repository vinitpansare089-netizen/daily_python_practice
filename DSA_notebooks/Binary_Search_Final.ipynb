```python
"""
Binary Search 
"""

nums = [-1,0,3,5,9,12]
target = 9

low = 0
mid = (low + high) // 2
high = len(nums) - 1

while low <= high:

    if nums[mid] == target:
        print(mid)

    elif target > nums[mid]:
        low = mid + 1

    else:
        high = mid - 1

print(-1)
```
