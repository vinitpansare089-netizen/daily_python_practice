```python
"""
Two Sum problem of hashmap
"""

nums = [3,2,4]
target = 6

hashmap = {}

for i, num in enumerate(nums):
    need = target - num
    if num in hashmap:
        print([hashmap[need], i])
        break

    hashmap[num] = i
    
print([hashmap[need], i])
```

```python
"""
is contains duplicate value or not solve using hashmap
"""
nums = [1,2,3,7]

seen = {}

for num in nums:
    if num in seen:
        seen[num] = True
        print(True)
        break

    seen[num] = True

else:
    print(False)
```

```python
"""
Valid Anagram using hashmap
"""
s = "anagram"
t = "nagaram"

freq1 = {}

for char in s:
    if char in freq1:
        freq1[char] += 1
    else:
        freq1[char] = 1

print(freq1)

freq2 = {}

for c in t:
    if c in freq2:
        freq2[c] += 1
    else:
        freq2[c] = 1

print(freq2) 

if freq1 == freq2:
    print(True)
else:
    print(False)
```

```python
"""
Ransom note using hashmap
"""
ransomNote = "aa"
magazine = "aab"

freq1 = {}
freq2 = {}

for char in ransomNote:
    if char in freq1:
        freq1[char] += 1
    else:
        freq1[char] = 1

freq2 = {}

for c in magazine:
    if c in freq2:
        freq2[c] += 1
    else:
        freq2[c] = 1

for char in freq1:
    if char not in freq1:
        print(False)
        break

    if freq2[char] < freq1[char]:
        print(False)
        break

else:
    print(True)
```
