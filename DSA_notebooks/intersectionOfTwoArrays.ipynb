```python
# num1 = [1,2,3,4,5]
# num2 = [2,3,6,4]

def twoArrays(num1: list, num2: list):

    for i in num1:
        for j in num2:
            if num1[i] == num2[j]:
                return list[i,j]
            else:
                return None
            
print(twoArrays(num1 = [1,2,3,4,5], num2 = [2,3,6,4]))

######### BAKWAS CODE ############
```

```python
def twoArrays(num1: list, num2:list) -> list:
    
    return list(set(num1) & set(num2))

print(twoArrays(num1=[1,2,3,4,5], num2 = [2,5,8,3]))
```
