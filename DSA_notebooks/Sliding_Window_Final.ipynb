```python
"""
Sliding window
"""

N = int(input())
arr = list(map(int, input().split()))
k = int(input())

print(arr)

window_sum = sum(arr[:k])
max_sum = window_sum

for i in range(k, len(arr)):
    window_sum -= arr[i - k]
    window_sum += arr[i]

    if window_sum > max_sum:

        max_sum = window_sum

print(max_sum)
```

```python

N = 6
arr = [2, 1, 5, 1, 3, 2]
target = 7

left = 0
window_sum = 0
answer = 0

for right in range(len(arr)):

    window_sum += arr[right]

    while window_sum > target:  
        window_sum -= arr[left]
        left += 1

    answer = max(answer, right - left +  1)

print(answer)
```

```python
"""
Minimum Sum in Sliding Window
"""

N = int(input())
arr = list(map(int, input().split()))
k = int(input())

window_sum = sum(arr[:k])
min_sum = window_sum

for i in range(k, len(arr)):
    window_sum -= arr[i - k]##Outgoing
    window_sum += arr[i]######Incoming

    if window_sum < min_sum:
        min_sum = window_sum

print(min_sum)
```
