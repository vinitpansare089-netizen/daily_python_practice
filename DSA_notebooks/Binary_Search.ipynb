```python
"""
Binary Search with three pointers
"""

N = int(input())
arr = list(map(int, input().split()))
target = int(input())

left = 0
right = len(arr) - 1


while left <=  right:

    mid = (left + right) // 2

    if arr[mid] == target:
        print(True)
        break

    elif arr[mid] < target:
        left = mid + 1
        
    else:
        right = mid - 1

else:
    print(False)
```
