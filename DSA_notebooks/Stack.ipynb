```python
"""
Stack operations
"""

stack = []

stack.append(10)
stack.append(20)
stack.append(30)
stack.append(40)

print(stack.pop())
print(stack.pop())

print(stack)
```

```python
"""
Brackets program 
"""

pairs = {
    ')': '(',
    ']': '[',
    '}': '{'
}

stack = []

for char in stack:

    if char in "([{":
        stack.append(char)

    else:
        if not stack:
            print(False)

        if stack.pop() != pairs[char]:
            print(False)
            break
else:
    print(len(stack) == 0)
```

```python
s = "((())"

stack = []

for char in s:
    if char == '(':
        stack.append(char)
    else:
        stack.pop()

print(len(stack) == 0)
```

```python
s = '(()())'

stack = []

for char in stack:
    if char in '(':
        stack.append(char)
    else:
        stack.pop()

print(len(stack) == 0)
```

```python
"""
Reverse string 
"""

s = str(input())
print(s)

stack = []

for char in s:
    stack.append(char)

while stack:
    print(stack.pop(), end ="")
```
