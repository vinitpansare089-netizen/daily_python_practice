```python
"""
A company gives me n numbers and i have to make it to return a maximum number
"""

N = int(input())
arr = list(map(int, input().split()))

#print(arr)
maximum = arr[0]

for num in arr:
    if num > maximum:
        maximum = num

print(maximum)
```

```python
"""
Sorting algorithms
"""

N = int(input())
arr = list(map(int, input().split()))

for i in range(N):
    min_index = i

    for j in range(i + 1, N):
        if arr[j] < arr[min_index]:
            min_index = j

    arr[i], arr[min_index] = arr[min_index], arr[i]

print(*arr)
```

```python
"""
taking multiple arrays 
"""
T = int(input())

for _ in range(T):

    N = int(input())
    arr = list(map(int, input().split()))

for i in range(N):
    min_index = i

    for j in range(i + 1, N):
        if arr[j] < arr[min_index]:
            min_index = j

    arr[i], arr[min_index] = arr[min_index], arr[i]

print(*arr)
```
