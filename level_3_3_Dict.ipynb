```python
from numbers import Number

def model_range(scores: dict) -> tuple:
  if not isinstance(scores, dict):
    raise TypeError("Inputs must be Dict")

  if len(scores) == 0:
    raise ValueError("Dict must not be empty")

  for k,v in scores.items():
    if not isinstance(k, str):
      raise TypeError("keys must be strings")

    if not isinstance(v, Number):
      raise TypeError("Values must be Numbers")

  best_key, best_value = next(iter(scores.items()))
  worst_key, worst_value = best_key, best_value

  for k,v in scores.items():
    if v > best_value:
      best_key, best_value = k,v

    if v < worst_value:
       worst_key, worst_value = k, v

  total = 0

  for v in scores.values():
    total = total + v

  average = total / len(scores)

  difference = best_value - worst_value
  return best_key, worst_key, difference , best_value, worst_value, average


scores = {
    "A": 10,
    "B": 30,
    "C": 25
}

print(model_range(scores))
```
