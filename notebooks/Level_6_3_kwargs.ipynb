```python
def my_info(**kwargs): 
    return kwargs

info = my_info(name="Vinit", age=23, add="Indore")

print(info)  
```

```python
def st_info(**kwargs):
    for k, v in kwargs.items():
        print(v) 

st_info(st1="vinit", st2="Harshit", st3="Abhi", st4="Ayush")
```

```python
# Product: Laptop | Price: 50000"
def product_info(**kwargs):
    result = []
    for k ,v in kwargs.items():
        result.append(f'{k.capitalize()}: {v}')
    return ' | '.join(result)
    
print(product_info(product= "Laptop", price=50000))
# lst = " | ".join(info) 
# print(info) 
```

```python
def product_info(**kwargs):
    return " | ".join(f'{k.capitalize()}: {v}' for k, v in kwargs.items())

info = (product_info(product= "Laptop", price=50000))
print(info)  
```
