```python
class Student:
    def __init__(self, name):
        #print("vinit")
        self.name = name
        
s1 = Student("Vinit")
s2 = Student("Mahi")

print(s1.name)
print(s2.name)
```

```python
class Student:
    def __init__(self, name):
        self.name = name

    def Say_hello(self):
        print("Hello, i am ",self.name)

s1 = Student("Vinit")
s1.Say_hello()
```

```python
class Car:
    def __init__(self, brand):
        self.brand = brand

    def show(self):
        print("The car brand is", self.brand)

s1 = Car("BMW")
s2 = Car("Audi")

s1.show()
s2.show()
```

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    students = [
        Student("vinit", 21),
        Student("mahi", 42)
    ]

    print(students)
 
```

```python
#Without class

print("Vinit",21)
print("mahi", 42)

#with class

class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("vinit", 21)
s2 = Student("Mahi", 42)

print(s1.name, s1.age)
print(s2.name, s2.age)
```

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show(self):
        print(self.name, self.age)

s1 = Student("Vinit", 21)
s2 = Student("Mahi", 23)

s1.show()
s2.show()
```
