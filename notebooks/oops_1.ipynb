```python
class Car:
    def __init__(self, brand):
        self.brand = brand

    def show(self):
        print("The car brand is", self.brand)

s1 = Car("BMW")
s2 = Car("Audi")

s1.show()
s2.show()
```

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

students = [
        Student("vinita", 21),
        Student("mahi", 42)
]

for s in students:
    print(s.name, s.age)

# print(students)
```
