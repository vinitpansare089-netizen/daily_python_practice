```python
def count(n):
    if n == 0:
        return 

    print(n)
    count(n - 1)

print(count(5))
```

```python
def factorial(n):
    if n == 0:
        return 1

    return n * factorial(n -  1)

print(factorial(4))
```

```python
result = 1

for i in range(1, 6):
    result *= i

print(result)
```
