```python
def count(n):
    if n == 0:
        return 

    print(n)
    count(n - 1)

print(count(5))
```

```python
def factorial(n):
    if n == 0:
        return 1

    return n * factorial(n -  1)

print(factorial(4))
```

```python
result = 1

for i in range(1, 6):
    result *= i

print(result)
```

```python
def print_numbers(n):

    if n == 0:
        return 

    print_numbers(n - 1)
    print(n)

print_numbers(5)
```

```python
def print_reverse(n):
    if n == 0:
        return 1

    print(n)
    print_reverse(n - 1)

print_reverse(5)
```

```python
def print_forward(n):
    if n == 0:
        return 

    print_forward(n - 1)
    print(n)

print_forward(5)
```

```python
def sum_numbers(n):
    if n == 0:
        return 0

    return n + sum_numbers(n - 1)

sum_numbers(5)
```

```python
def factorial_num(n):
    if n == 0:
        return 1

    return n * factorial_num(n - 1)

factorial_num(5)
```

```python
def power_num(n, power):
    if n == 0:
        return 1


    return power_num(n ** power)

power_num(2,3)

    
```
