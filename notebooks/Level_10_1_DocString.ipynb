```python
# Write:

# def calculate_avg(marks):

# Add:

# docstring
# return average

def calculate_avg(marks: list[int]) -> float:

    """ 
            takes marks as an list haing all int element
    """

    if not all(isinstance(x, int) for x in marks):
        raise TypeError('Only int allowed')

    return sum(marks) / len(marks)

# print(calculate_avg([1,2.9,3.3,4.5]))
print(calculate_avg([1,2,3,4,5]))


print(calculate_avg.__doc__)
```
