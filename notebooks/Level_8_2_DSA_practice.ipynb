```python
def max_function(nums):
    mx = nums[0]

    for num in nums:
        if num > mx:
            mx = num
    return mx

nums = [100,22,44,6,7,4]
print(max_function(nums))
```

```python
def duplicate(nums):
    temp = nums[0]

    for num in nums:
        if num == temp:
           temp = num.split()
    return temp

nums = ['ed   rhoham   ']
print(duplicate(nums))

```

```python
def duplicate(nums):
    result = []

    for num in nums:
        if num not in result:
            result.append(num)
    return result

nums = [1,1,1,2,2,3,3,4,5,6,6,6]
print(duplicate(nums))
```

```python
# nums = {"vinit"}/// this is set vinit not dictionary
word = "banana"

def frequency(word):
    result = {}

    for char in word:
        if char in result:
            result[char] += 1
        else:
            result[char] = 1
    return result

print(frequency(word))
```
