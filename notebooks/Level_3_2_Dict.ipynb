```python
from numbers import Number

def worst_model(scores: dict) -> tuple:
  if not isinstance(scores, dict):
    raise TypeError("Input must be Dict")

  if len(scores) == 0:
    raise ValueError("Dict must not be empty")

  for k,v in scores.items():
    if not isinstance(k, str):
      raise TypeError("Keys must be strings")

    if not isinstance(v, Number):
      raise TypeError("Values must be numbers")

  worst_key = None
  worst_value = None

  for k, v in scores.items():
    if worst_value is None or v < worst_value:
      worst_key = k
      worst_value = v

  return worst_key , worst_value

scores = {
    "model_1": 72.1,
    "model_2": 60.2,
    "model_3": 91.3
}

print(worst_model(scores))
```
