```python
#AREA OF CIRCLE

def area_circle(r: float):
  pi = 3.14
  A = pi * (r * r)
  return A

area_circle(6)

#MAX OF THREE NUMBERS

def max_of_three(x: float, y: float, z: float):
  if x>y and x>z:
    return x, "x"
  elif y>z and y>x :
    return y, "y"
  else:
    return z, "z"

value, name = max_of_three(5.23, 4.54, 8.35)

print(value ,name)


#CONVERT CALCIUS TO FEHRENHEIT


def to_fehrenheit(c: float):
  return (c * 1.8) + 32

to_fehrenheit(45)

```
