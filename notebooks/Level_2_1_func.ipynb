```python
def safe_mean(numbers: list) -> float:

  if not isinstance(numbers, list):
    raise TypeError("Inputs must be list")

  if len(numbers) == 0:
    raise ValueError("List mush not be empty")

  for n in numbers:
    if not isinstance(n, (int, float)):
      raise TypeError("All elements must be Numbers")

  return sum(numbers) /  len(numbers)

print(safe_mean([1,2,3,4,5]))
```

```python
# Rules

# Must be list
#  Not empty
#  All numbers
#  Return sum

def safe_sum(numbers: list) -> float:
  if not isinstance(numbers, list):
    raise TypeError("Must be a List")

  if len(numbers) == 0:
    raise ValueError("List must not be empty")

  for n in numbers:
    if not isinstance(n, (int, float)):
      raise ValueError("List must contains numbers")

  return sum(numbers)

print(safe_sum([1,2,3,4]))
```

```python
from numbers import Number

def safe_max(numbers: list) -> float:

  if not isinstance(numbers, list):
    raise ValueError("Input must be List")

  if len(numbers) == 0:
    raise ValueError("List must not be empty")

  for n in numbers:
    if not isinstance(n, Number):
      raise TypeError("input must be numbers")


#returns Maximum values
  current_max = numbers[0]

  for n in numbers:
    if n > current_max:
      current_max = n

  # check value if come in list more than 1 times
  # if numbers.count(current_max) > 1:
  #   raise ValueError("Appears more time one number")


  #Return Indexes of repeating value
  indexes = []

  for i, n in enumerate(numbers):
    if n == current_max:
      indexes.append(i)


  return indexes

  return current_max

print(safe_max([1, 9, 4, 2]))
print(safe_max([1, 9, 4, 9]))
```
