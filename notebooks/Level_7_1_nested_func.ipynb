```python
def outer(x):
    def inner(y):
        return y * 2
    return inner(x)

print(outer(4))#//8
```

```python
def num1(x):
    def num2(y):
        return y ** 2
    return num2(x)


print(num1(6))
```

```python
def apply_func(func, value):
    return func(value)

def square(x):
    return x * x

print(apply_func(square, 5))
```

```python
def apply_function(func, x):
    return func(x)

def cube(x):
    return x * x * x

print(apply_function(cube, 5))
```

```python
# Q2:

# Create 2 functions:

# double(x)
# square(x)

def apply_func(func, x):
    return func(x)

def double(x):
    return x + x

def square(x):
    return x * x

print(apply_func(double, 5))
print(apply_func(square, 5))
```
