```python
nums = [1,2,3,4,5,6,7,8]

result = filter(lambda x: x % 2 != 0, nums)

print(list(result))
```

```python
word = ["vinit", "harshit", "abhi", "ayush"]

result = filter(lambda x: len(x) > 5 , word)

print(list(result))
```

```python
nums = [1,2,3,4,5,6,7,8]

# result = filter(lambda x: x % 2 == 0, nums)

def is_even(x):
    return x % 2 != 0

result = filter(is_even, nums)


print(list(result))
```
