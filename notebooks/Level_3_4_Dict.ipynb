```python
from numbers import Number

def top_k_models(scores: dict, k: int) -> list:

    if not isinstance(scores, dict):
        raise TypeError("Input must be dict")

    if len(scores) == 0:
        raise ValueError("Dict must not be empty")

    if not isinstance(k, int) or k <= 0:
        raise ValueError("k must be positive integer")

    if k > len(scores):
        raise ValueError("k cannot be greater than number of models")

    for key, value in scores.items():
        if not isinstance(key, str):
            raise TypeError("Keys must be strings")
        if not isinstance(value, Number):
            raise TypeError("Values must be numbers")

    result = []
    temp = scores.copy()

    for _ in range(k):

        best_key, best_value = next(iter(temp.items()))

        for key, value in temp.items():
            if value > best_value:
                best_key, best_value = key, value

        result.append((best_key, best_value))
        del temp[best_key]

    return result


scores = {
    "A": 10,
    "B": 30,
    "C": 65
}

print(top_k_models(scores, 2))
```
