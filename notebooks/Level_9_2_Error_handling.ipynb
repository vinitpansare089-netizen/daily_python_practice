```python
def safe_divide(a,b):
    try:
        return a / b
    
    except ZeroDivisionError:
        return "Invalid division"

print(safe_divide(10 ,0))
```

```python
def safe_int(x):
    try:
        return int(x)
    
    except ValueError:
        return 'Invalid'
    
print(safe_int('abc'))##Invalid
print(safe_int(5))##5
```

```python
def convert_and_divide(x, y):
    try:
        return int(x) / y
    
    except ValueError:
        return "Invalid"
    
    except ZeroDivisionError:
        return "cannot divide by zero"

print(convert_and_divide("10", 5))###2.0
print(convert_and_divide("abc", 2))##Invalid
# print(convert_and_divide("10", 0))
```
