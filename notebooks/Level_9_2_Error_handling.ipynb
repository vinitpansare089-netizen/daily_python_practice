```python
def safe_divide(a,b):
    try:
        return a / b
    
    except ZeroDivisionError:
        return "Invalid division"
    
print(safe_divide(10 ,0))
```

```python
def safe_int(x):
    try:
        x = int(x)
        return x
    
    except TypeError:
        return 'Not Valid'
    
print(safe_int('abc'))
```
