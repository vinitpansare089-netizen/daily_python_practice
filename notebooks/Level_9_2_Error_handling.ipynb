```python
def safe_divide(a,b):
    try:
        return a / b
    
    except ZeroDivisionError:
        return "Invalid division"
    
print(safe_divide(10 ,0))
```

```python
def safe_int(x):
    try:
        return int(x)
    
    except ValueError:
        return 'Invalid'
    
print(safe_int('abc'))
print(safe_int(5))
```
