```python
def max_function(nums):
    mx = nums[0]

    for num in nums:
        if num > mx:
            mx = num
    return mx

nums = [1,22,44,6,7,4]
print(max_function(nums))
```

```python

def max_function(*args):
    return max(args)

result = max_function(1,2,3,44,55)
max_function(result)
```
