```python

def max_function(nums):
    mx=nums[0]
    for lst in nums:
         if(nums[lst]>mx):
              temp =mx
              mx=nums[lst]
              nums[lst]=temp
    return  mx

nums = [10,20,30,40]

print(max_function(nums))

    
```
