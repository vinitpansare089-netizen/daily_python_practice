```python
"""
Write a function that counts how many vowels are in a string.
"""
def count_vowels(text):

    count = 0

    for char in text:

        if char in 'aeiou':
            count += 1

    return count

print(count_vowels("programming"))
```

```python
"""
Character Frequency
"""
def char_frequency(text):

    result = {}

    for char in text:

        if char in result:
            result[char] += 1
        else:
            result[char] = 1

    duplicates = {}

    for char, count in result.items():

        if count > 1:
            duplicates[char] = count

    return duplicates


print(char_frequency("hello"))
```

```python
"""
First Non-Repeating Character
"""
def first_unique(text):

    result = {}
    count = 0

    for char in text:

        if char in result:
            result[char] += 1
        else:
            result[char] = 1


    for char,count in result.items():

        if count == 1:
            return char

    return False

print(first_unique("aabbcdd"))
```

```python
"""
Palindrome Check
"""

def is_palindrome(text):

    left = 0
    right = len(text) - 1

    while left < right:

        if text[left] != text[right]:
            return False

        left += 1
        right -= 1

    return True

print(is_palindrome('vinit'))
print(is_palindrome('madam'))
```

```python
"""
Remove Duplicate Characters
"""

def remove_duplicates(text):

    seen = set()
    result = ''

    for char in text:
        if char not in seen:
            result += char
            seen.add(char)
    return result

print(remove_duplicates("programming"))
```
