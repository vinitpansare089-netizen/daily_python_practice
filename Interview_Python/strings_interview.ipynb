```python
# Length
# First character
# Last character
# Lowercase version
# Uppercase version
# Remove the spaces from beginning/end

text = "  Python Programming  "

print(len(text))
print(text[0])
print(text[-1])
print(text.upper().strip())
print(text.strip())
```

```python
"""
String Indexing & Slicing
"""

text = "Python Programming"

print(text[0]) #//p
print(text[6]) #// ''
print(text[-1]) #//g
print(text[0:6]) #// Python
print(text[7:]) #// Programming
print(text[::-1]) #// gnimmargorP nohtyP
```
