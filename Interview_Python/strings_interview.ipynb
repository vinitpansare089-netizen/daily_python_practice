```python
"""
Write a function that counts how many vowels are in a string.
"""
def count_vowels(text):

    count = 0

    for char in text:

        if char in 'aeiou':
            count += 1

    return count

print(count_vowels("programming"))
```

```python
"""
Character Frequency
"""
def char_frequency(text):

    result = {}

    for char in text:

        if char in result:
            result[char] += 1
        else:
            result[char] = 1

    duplicates = {}

    for char, count in result.items():

        if count > 1:
            duplicates[char] = count

    return duplicates

print(char_frequency("hello"))
```

```python
"""
First Non-Repeating Character
"""
def first_unique(text):

    result = {}
    count = 0

    for char in text:

        if char in result:
            result[char] += 1
        else:
            result[char] = 1

    for char,count in result.items():

        if count == 1:
            return char

    return False

print(first_unique("aabbcdd"))
```

```python
"""
Palindrome Check
"""
def is_palindrome(text):

    left = 0
    right = len(text) - 1

    while left < right:

        if text[left] != text[right]:
            return False

        left += 1
        right -= 1

    return True

print(is_palindrome('vinit'))
print(is_palindrome('madam'))
```

```python
"""
Remove Duplicate Characters
"""

def remove_duplicates(text):

    seen = set()
    result = ''

    for char in text:
        if char not in seen:
            result += char
            seen.add(char)
    return result

print(remove_duplicates("programming"))
```

```python
"""
Reverse the string
"""
def reverse_string(text):

    s = list(text)
    left = 0
    right = len(text) - 1

    while left < right:

        s[left], s[right] = s[right], s[left]

        left += 1
        right -= 1

    return ''.join(s)


print(reverse_string('vinit'))
```

```python
"""
Anagram Problem
"""
def is_anagram(s1, s2):

    f1 = {}
    f2 = {}
    
    for char in s1:
        if char in f1:
            f1[char] += 1
        else:
            f1[char] = 1

    for char in s2:
        if char in f2:
            f2[char] += 1
        else:
            f2[char] = 1

    return f1 == f2


print(is_anagram('Vinit', 'tanvi'))
print(is_anagram("listen", "silent"))


```

```python
"""
Remove Spaces
"""
def remove_spaces(text):

    t = ''

    for char in text:

        if char != ' ':
            t += char

    return t

print(remove_spaces("VInit is a good boy"))

```

```python
"""Check if a String Contains Only Digits"""

def is_only_digits(text):

    for char in text:
        if not char.isdigit():
            return False
    return True
            

print(is_only_digits("1235"))
print(is_only_digits("123we5"))
```
