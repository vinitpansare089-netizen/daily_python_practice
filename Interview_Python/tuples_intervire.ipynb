```python
"""
First element
Last element
Number of times 20 occurs
Index of 40
"""

t = (10, 20, 30, 20, 40, 20)

print(t[0])#// 10
print(t[-1]) #// 20
print(t.count(20)) #// 3
print(t.index(40)) #// 4

print(t.index(20)) #// 1
```

```python
"""
Packing and unpacking
"""

student = ("vinit", 22, "MCA")

name, age, course = student

print(name)
print(age)
print(course)
```

```python
"""
Extended Unpacking
"""

data = (5, 10, 15, 20, 25, 30, 35)

first, *middle, last = data

print(first)
print(middle)
print(last)
```

```python
"""
Nested Tuples
"""

students = (
    ("vinit", 22),
    ("Rahul", 23),
    ("Yashi", 21),
)

print(students[0][0])
print(students[0][1])
print(students[2][0])
print(students[2][1])
```

```python
"""
Tuple Converions
"""
nums = [1, 2, 3]

t = tuple(nums)

print(t)
```

```python
"""
list to tuple and tuple to list conversion
"""
data = [10, 20, 30, 40]

t = tuple(data)
print(t)

nums = list(t)
print(nums)
```

```python
"""
Count Occurance in tuples
"""

def count_value(t, target):

    count = 0

    for num in t:
        if num == target:
            count += 1

    return count

print(count_value((10, 20, 10, 30, 10, 40), 10))
```

```python
"""
Find maximum value
"""

def max_tuple(t):

    max = t[0]

    for num in t:
        if num > max:
            max = num
    return max

print(max_tuple((10, 5, 27, 3, 19)))
print(max_tuple((-10, -5, -20)))
```
