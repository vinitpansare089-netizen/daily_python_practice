```python
student = {
    "name": "Vinit",
    "age": 22,
    "course": "MCA"
}

print(student['name'])
student['age'] = 23
student['city'] = 'Indore'
del student['course']
print(student)
```

```python
student = {
    "name": "Vinit",
    "age": 22,
    "course": "MCA"
}

print(student.keys())
print(student.values())
print(student.items())
print(student.get('name'))
print(student.get('city', 'unknown'))
```
