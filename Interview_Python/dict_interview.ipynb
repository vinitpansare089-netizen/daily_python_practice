```python
"""
Looping Through Dictionaries
"""
def print_student(student):

    for key, value in student.items():
        print(key,value)


print(print_student({
    "name": "Vinit",
    "age": 23,
    "course": "MCA"
}))
```

```python
# Dictionary Comprehension

nums = [1, 2, 3, 4, 5]

squares = {num: num * num for num in nums}

print(squares)
```

```python
def frequency(nums):

    count = {}

    for num in nums:

        if num in count:
            count[num] += 1
        else:
            count[num] = 1

    return count

print(frequency([1, 2, 2, 3, 1, 2, 4]))
```

```python
def frequent_element(nums):

    most_frequent = None
    highest_count = 0
    count = {}

    for num in nums:

        if num in count:
            count[num] += 1
        else:
            count[num] = 1

    for num, freq in count.items():

        if freq > highest_count:
            highest_count = freq
            most_frequent = num

    return most_frequent


print(frequent_element([1, 2, 2, 3, 1, 2, 4]))
```

```python
def reverse_dict(data):

    result = {}

    for key, value in data.items():
        result[value] = key

    return result


print(reverse_dict({
    "a": 1,
    "b": 2,
    "c": 3
}))
```

```python
def high_marks(marks):

    result = {}

    for key, value in marks.items():

        if value > 80:
            result[key] = value

    return result

print(high_marks({
    "Math": 85,
    "Python": 92,
    "DBMS": 76,
    "ML": 95
}))
```

```python
def count_names(names):

    result = {}


    for char in names:

        if char in result:
            result[char] += 1
        else:
            result[char] = 1

    return result

print(count_names(["Vinit", "Rahul", "Vinit", "Aman", "Rahul", "Vinit"]))
```
