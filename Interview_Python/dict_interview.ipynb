```python
student = {
    "name": "Vinit",
    "age": 22,
    "course": "MCA"
}

print(student['name'])
student['age'] = 23
student['city'] = 'Indore'
del student['course']
print(student)
```

```python
student = {
    "name": "Vinit",
    "age": 22,
    "course": "MCA"
}

print(student.keys())
print(student.values())
print(student.items())
print(student.get('name'))
print(student.get('city', 'unknown'))
```

```python
"""
Looping Through Dictionaries
"""
def print_student(student):

    for key, value in student.items():
        print(key,value)


print(print_student({
    "name": "Vinit",
    "age": 23,
    "course": "MCA"
}))
```

```python
students = {
    "student1": {
        "name": "Vinit",
        "age": 23
    },
    "student2": {
        "name": "Rahul",
        "age": 22
    }
}

print(students["student1"]['name'])
print(students["student2"]['name'])
print(students["student2"]['age'])
```

```python
# Dictionary Comprehension

nums = [1, 2, 3, 4, 5]

squares = {num: num * num for num in nums}

print(squares)
```

```python
def frequency(nums):

    count = {}

    for num in nums:

        if num in count:
            count[num] += 1
        else:
            count[num] = 1

    return count

print(frequency([1, 2, 2, 3, 1, 2, 4]))
```
