```python
"""Write a Python program that:
1. Creates a file called practice.txt and writes Python is powerful into it.
2. Appends I am learning file handling on a new line.
3. Reads the entire file and prints its contents."""

with open('practice.txt', 'w') as file:
    file.write("Python is powerful")

with open('practice.txt', 'a') as file:
    file.write('\nI am learning file handling')

with open('practice.txt', 'a') as file:
    file.write('\nMachine Learning AI')

with open('practice.txt', 'r') as file:
    content = file.read()

print(content)
```

```python
with open("practice.txt", "r") as file:
    first = file.readline()
    second = file.readline()
    all_content = file.read()

print("First:", repr(first))
print("Second:", repr(second))
print("Remaining:", repr(all_content))
```

```python
with open("practice.txt", "r") as file:
    print(file.readline())
    file.seek(0)
    print(file.readline())
```

```python
with open("practice.txt", "r") as file:
    file.seek(2)
    print(file.read(2))
    print(file.tell())
```
