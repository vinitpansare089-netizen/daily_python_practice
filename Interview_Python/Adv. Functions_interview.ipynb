```python
"""
# find the largest number WITHOUT using max()
"""

def find_max(*args):

    max_value = args[0]

    for num in args[1:]:
        if num > max_value:
            max_value = num

    return max_value

print(find_max(4, 9, 2, 15, 7))
print(find_max(-5, -2, -10))
```

```python
"""
print each key and value
"""

def student_info(**kwargs):

    for key,value in kwargs.items():
        print(key, ':',value)

student_info(name="Vinit", course="MCA", year=2026)
```

```python
def profile(name, *skills, **details):

    print(f'name: {name}')
    
    for i in skills:
        print(f'skills: {i}')

    for key,value in details.items():
        print(key, ':', value)

profile(
    "Vinit",
    "Python",
    "SQL",
    "PyTorch",
    college="Medi-Caps",
    year=2026
)
```

```python
x = 5

def change():
    global x
    x = x + 1
    print(x)

change()
```

```python
def outer():
    
    x = 10

    def inner():
        nonlocal x
        x += 1

    inner()
    print(x)

outer()
```

```python
"""
lembda functions vs normal function
"""

def squares(x):
    return x ** 2

print(squares(7))

square = lambda x : x**2

print(square(7))
```

```python
nums = [1, 2, 3, 4, 5]

result = map(lambda x : x * 2, nums)

print(list(result))
```

```python
nums = [1, 2, 3, 4, 5, 6, 7, 8]
 
even = filter(lambda x: x % 2 == 0, nums)

print(list(even))
```

```python
from functools import reduce

nums = [1, 2, 3, 4, 5]

redu = reduce(lambda x,y: x * y, nums)

print(redu)
```
