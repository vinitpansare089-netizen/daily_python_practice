```python
"""
Classes and objects
"""

class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"My name is {self.name} and I am {self.age} years old.")


student1 = Student("Alice", 20)
student2 = Student("Bob", 21)

student1.introduce()
student2.introduce()
```

```python
class student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f'My name is {self.name} and my age is {self.age}')

student1 = student('vinit', 22)
student2 = student('yashi', 22)

# student1.introduce(student2) // Type Error positional arguments

student1.introduce()
student2.introduce()
```
