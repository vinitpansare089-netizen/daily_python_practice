```python
"""
Recursive function for sum
"""

def recursive_sum(n):

    if n == 0:
        return 0

    return n + recursive_sum(n - 1)

print(recursive_sum(5))
```

```python
"""
Sum of the digits 
"""

def digit_sum(n):

    if n == 0:
        return 0

    return n % 10 + digit_sum(n // 10)

print(digit_sum(12345))
    
```

```python
"""Reverse a number recursively"""

def reverse_number(n, result=0):

    if n == 0:
        return result

    return reverse_number(n // 10, result * 10 + n % 10)

print(reverse_number(1234))
```

```python
"""Recursive power"""

def power(base, exponent):

    if exponent == 0:
        return 1

    return base * power(base, exponent - 1)

print(power(5, 2))
```

```python
"""Fibonacci"""

def fibonacci(n):

    if n == 0:
        return 0

    if n == 1:
        return 1

    return  fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(8))

"""Fibonacci 2nd method"""

def fibonacci1(n):

    if n == 0 or n == 1:
        return n

    return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(7))
```

```python
"""Count Digits Recursively"""

def count_digits(n, count = 0):

    if n == 0:
        return count

    return count_digits(n // 10, count + 1)

print(count_digits(43654))
```

```python
"""Reverse a String Recursively"""

def reverse_string(text):

    if text == '':
        return ''

    return reverse_string(text[1:]) + text[0]

print(reverse_string('vinit'))
```

```python
"""Find Maximum in a List Recursively"""

def recursive_max(nums):

    if len(nums) == 1:
        return nums[0]

    rest_max = recursive_max(nums[1:])

    if nums[0] > rest_max:
        return nums[0]

    else:
        return rest_max

print(recursive_max([3, 7, 2, 9, 5]))
```

```python
"""
Test question : factorial
"""
def factorial(n):
    if n == 0:
        return 1

    return n * factorial(n - 1)

print(factorial(5))
```
