```python
"""
Classes and objects
"""

class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"My name is {self.name} and I am {self.age} years old.")


student1 = Student("Alice", 20)
student2 = Student("Bob", 21)

student1.introduce()
student2.introduce()
```

```python
"""
Requirements:
1. Create a class named Car.
2. Use __init__ to receive brand and color.
3. Store them as instance attributes using self.
4. Create a method named show_details() that prints the brand and color.
5. Create two objects with different values and call the method on both.
"""

class Car:

    def __init__(self, brand, color):
        self.brand = brand
        self.color = color

    def show_details(self):
            print(f'the car brand name is {self.brand} and color is {self.color}')

c1 = Car('BMW', 'Blue')
c2 = Car('toyota', 'white')

c1.show_details()
c2.show_details()

    
```
