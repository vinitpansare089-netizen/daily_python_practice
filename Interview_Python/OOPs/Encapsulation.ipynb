```python
""" GETTER AND SETTER METHODS  
Create an employee with a salary of 50000.
Print the salary using get_salary().
Change it to 60000 using set_salary().
Attempt to change it to -1000.
Print the salary again to confirm that the invalid update was rejected.
"""

class Employee:

    def __init__(self, salary):
        self.__salary =  salary

    def get_salary(self):
        return self.__salary

    def set_salary(self, new_value):
        if new_value > 0:
            self.__salary = new_value
        else:
            print('Salary must be positive')

obj = Employee(50000)
print(obj.get_salary())

obj.set_salary(100000)
print(obj.get_salary())

obj.set_salary(-5000)
print(obj.get_salary())
```

```python
"""
@property controll attributes
"""

class Employee:

    def __init__(self, salary):
        self.__salary =  salary

    @property
    def salary(self):
        return self.__salary
    
    @salary.setter
    def salary(self, new_value):
        if new_value > 0:
            self.__salary = new_value
        else:
            print('Salary must be positive')


obj = Employee(50000)
print(obj.salary)
obj.salary = 600000
print(obj.salary)
obj.salary = -50000
print(obj.salary)
```

```python
"""
Build a BankAccount class. This is a practical exercise that combines private attributes, methods, and validation.
Requirements:
1. In __init__, accept owner and balance.
2. Store owner as a public attribute.
3. Store balance as a private-style attribute, self.__balance.
4. Use @property to expose the balance through account.balance.
5. Add a property setter that rejects negative balances.
6. Add a deposit(amount) method that accepts only positive amounts and increases the balance.
7. Add a withdraw(amount) method that withdraws money only when the amount is positive and sufficient funds are available.
8. Test a successful deposit, a successful withdrawal, and an invalid transaction.
"""

class BankAccount:

    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self,new_balance):
        if new_balance > 0:
            self.__balance = new_balance
        else:
            print('Balance must be Positive')

    def deposit(self,amount):
        if amount > 0:
            self.__balance += amount
        else:
            print('amount must be positive')

    def withdraw(self,amount):
        if amount <= 0:
            print('withdrawl must be positive')
        elif amount > self.__balance:
            print('Insuffient Balance')
        else:
            self.__balance -= amount


acc = BankAccount('vinit', 1000000)

acc.deposit(500)
acc.withdraw(300)
print(acc.balance)

acc.withdraw(500000000)
print(acc.balance)
```
