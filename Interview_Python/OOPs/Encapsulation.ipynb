```python
"""PUBLIC ATTRIBUTES
Encapsulation

In this Program how easy to access and manipulate the salary but not using ecapsulation
"""
class Employee:

    def __init__(self, salary):
        self.salary = salary

s1 = Employee(50000)
s1.salary

s1.salary = 1000000 #// public attributes
s1.salary
```

```python
""" PRIVATE ATTRIBUTES
Modify the Employee class so that:
1. It receives salary in __init__.
2. It stores the value as self.__salary.
3. It has a method called show_salary() that prints the salary.
4. Create an employee with a salary of 50000 and call show_salary().
"""

class Employee:

    def __init__(self,salary):
        self.__salary = salary

    def show_salary(self):
        print(self.__salary)

e1 = Employee(100000)
e1.__salary = 500
e1.show_salary()
```

```python
""" GETTER AND SETTER METHODS  
Create an employee with a salary of 50000.
Print the salary using get_salary().
Change it to 60000 using set_salary().
Attempt to change it to -1000.
Print the salary again to confirm that the invalid update was rejected.
"""

class Employee:

    def __init__(self, salary):
        self.__salary =  salary

    def get_salary(self):
        return self.__salary

    def set_salary(self, new_value):
        if new_value > 0:
            self.__salary = new_value
        else:
            print('Salary must be positive')

obj = Employee(50000)
print(obj.get_salary())

obj.set_salary(100000)
print(obj.get_salary())

obj.set_salary(-5000)
print(obj.get_salary())
```
