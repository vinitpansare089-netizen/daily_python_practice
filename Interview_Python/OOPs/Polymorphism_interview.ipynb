```python
"""
polymorphism : means same method name produce different behavior depending on the object 
"""
"""
Create a Car class with a method start() that prints "Car engine started".
Create a Bike class with a method start() that prints "Bike engine started".
Create one object of each class.
Call start() on both objects.
"""

class Car:
    def start(self):
        print("Car engine started")

class Bike:
    def start(self):
        print('Bike engine started')

car = Car()
bike = Bike()

car.start()
bike.start()
```

```python
"""
Build your own version using a parent class named Notification.
- Notification has a method send() that prints "Sending a notification".
- EmailNotification inherits from Notification and overrides send() to print "Sending an email".
- SMSNotification inherits from Notification and overrides send() to print "Sending an SMS".
- Create both child objects and call send() on each.
"""

class Notification:
    def send(self):
        print('Sending a notification')

class EmailNotification(Notification):
    def send(self):
        print('Sending an email')

class SMSNotification(Notification):
    def send(self):
        print('Sending an sms')

email = EmailNotification()
sms = SMSNotification()

email.send()
sms.send()
```
