```python
"""
Create:
- A parent class named Animal.
- An attribute called name.
- A method called make_sound() that prints "Animal makes a sound".
- A child class named Dog that inherits from Animal.
- Create a Dog object named "Bruno" and call make_sound().
"""

class Animal:

    def __init__(self, name):
        self.name = name

    def make_sound(self):
        print(f'{self.name} makes a sound')

class Dog(Animal):

    def bark(self):
        print(f'{self.name} says woof!')

d1 = Dog('Bruno')
d1.make_sound()
d1.bark()
```

```python
class Animal:

    def __init__(self, name, age, color):
        self.name = name
        self.age = age
        self.color = color

class Dog(Animal):

    def __init__(self,name, age, color, breed):
        super().__init__(name,age,color)
        self.breed = breed

    def identity(self):
        print(f'the Dog name is {self.name} age is {self.age} color is {self.color} and breed is {self.breed}')


d1 = Dog('Bruno', 22, 'Black', 'German sheferd')
d1.identity()
```

```python
""" ---MULTILEVEL INHERITANCE----
Create this inheritance chain yourself:
- Person has a method introduce() that prints "I am a person".
- Student inherits from Person and adds a method study() that prints "I am studying".
- CollegeStudent inherits from Student and adds a method attend_class() that prints "Attending class".
- Create a CollegeStudent object and call all three methods.
"""

class Person:

    def introduce(self):
        print('I am a person')

class Student(Person):

    def study(self):
        print('I am studying')

class CollegeStudent(Student):

    def attend_class(self):
        print('Attending class')

c1 = CollegeStudent()
c2 = Student()
c1.introduce()
c1.study()
c1.attend_class()
# c2.attend_class() /// not run
```
