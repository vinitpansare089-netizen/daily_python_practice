```python
"""
Previously Classes and objects..without inheritance we had to make many classes and repeating codes
"""

class Student:

    def __init__(self, name):
        self.name = name

    def introduce(self):
        print(f'my name is {self.name}')


class Teacher:

    def __init__(self, name):
        self.name = name

    def introduce(self):
        print(f'my name is {self.name}')


s1 = Student('vinit')
s1.introduce()
s1 = Teacher('Yashi')
s1.introduce()
```

```python
"""
Create:
- A parent class named Animal.
- An attribute called name.
- A method called make_sound() that prints "Animal makes a sound".
- A child class named Dog that inherits from Animal.
- Create a Dog object named "Bruno" and call make_sound().
"""

class Animal:

    def __init__(self, name):
        self.name = name

    def make_sound(self):
        print(f'{self.name} makes a sound')

class Dog(Animal):

    def bark(self):
        print(f'{self.name} says woof!')

d1 = Dog('Bruno')
d1.make_sound()
d1.bark()
```
