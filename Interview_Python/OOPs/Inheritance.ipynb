```python
"""
Hybrid Inheritance
"""

class person:
    def introduce(self):
        print('I am a person')

class Student(person):
    def study(self):
        print("I am studying")

class Employee(person):
      def work(self):
        print("I am working")

class WorkingStudent(Student, Employee):
    pass

ws = WorkingStudent()
e = Employee()

e.introduce()

ws.introduce()
ws.study()
ws.work()

print(WorkingStudent.mro())
```

```python
"""
Challenge: Employee management system
Write a program with these requirements:
1. Create a Person class with a constructor that stores name.
2. Create an Employee class that inherits from Person and adds salary.
3. Create a Developer class that inherits from Employee and adds programming_language.
4. Use super() to initialize the parent attributes instead of repeating their assignments.
5. Add a show_details() method to Developer that prints all three attributes.
6. Create one developer and call show_details().
"""
class Person:
    def __init__(self, name):
        self.name = name

class Employee(Person):

    def __init__(self,name,salary):
        super().__init__(name)
        self.salary = salary

class Developer(Employee):

    def __init__(self, salary,name, programming_language):
        super().__init__(salary,name)
        self.programming_language = programming_language

    def show_details(self):
        print(f'my name is {self.name} with {self.salary} salary and langusage is {self.programming_language}')
    
dev = Developer('vinit', 1000000, 'Python')

dev.show_details()
```
