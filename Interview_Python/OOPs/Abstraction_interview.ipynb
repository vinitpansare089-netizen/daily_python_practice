```python
"""
Abstraction in python
Try creating a Notification abstract class with an abstract method send(self).
Then create an EmailNotification child class that implements send() and prints "Email sent successfully".
"""
from abc import ABC, abstractmethod

class Notification(ABC):

    def send(self):
        pass

class EmailNotification(Notification):
    def send(self):
        print('Email sent successfully!')

obj = EmailNotification()
obj.send()



```
