```python
"""
set in python
"""

nums = {10,20,30}

nums.add(40)
print(nums)
nums.remove(30)
print(nums)
nums.discard(50)
print(nums)
```

```python
"""
Union in sets
"""

A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A | B)
print(A & B)
print(A - B)
print(B - A)
print(A ^ B)
```

```python
"""
Set memberships
"""
nums = {10, 20, 30, 40}

print(20 in nums)
print(50 not in nums)
```

```python
A = {1, 2, 3}
B = {1, 2}

print(B.issubset(A))
print(A.issuperset(B))
print(A.isdisjoint({5,6}))
print(5 in A)
```

```python
"""
Interview questions
"""

def remove_duplicates(nums):

    return set(nums)

print(remove_duplicates([1, 2, 2, 3, 4, 4, 5, 5]))
```

```python
"""
Common Elements
"""

def common_elements(a:set, b:set):

    return set(a) & set(b)

print(common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))
```

```python
"""
set difference
"""

def difference(a, b):

    return set(a) - set(b)

difference([1, 2, 3, 4, 5],[3, 4, 5, 6, 7])
```

```python
"""
unique elements
"""
def unique_elements(a, b):

    return set(a) ^ set(b)

unique_elements([1, 2, 3, 4],[3, 4, 5, 6])
```

```python
"""Common Unique Values"""

def count_unique(nums):

    return len(set(nums))

print(count_unique([1, 2, 2, 3, 4, 4, 5, 6, 6]))
```

```python
"""
Has duplicate values??
"""

def has_duplicate(nums):

    seen = set()

    for num in nums:
        if num in seen:
            return True

        seen.add(num)

    return False

has_duplicate([1, 2, 3, 2, 4, 1, 5, 3])
```
