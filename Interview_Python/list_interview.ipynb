```python
"""
max,min number 
"""

def max_num(nums: list):

    max = nums[0]

    for num in nums:
        if num > max:
            max = num

    return max

print(max_num([-5, -2, -10, -1]))
print(max_num([10, 5, 27, 3, 19]))
print(max_num([7]))
```

```python
"""
Min Value
"""

def min_nums(nums:list) -> list:

    # min = nums[0]////Indexerror it will give

    if not nums:
        return None
    
    min = nums[0]

    for num in nums:
        if num < min:
            min = num

    return min

print(min_nums([10, 5, 27, 3, 19]))
print(min_nums([-5, -2, -10, -1]))
print(min_nums([]))
```

```python
"""
Count Occurance
"""

def count_occurances(nums, target):

    count = 0

    for num in nums:
        if num == target:
            count += 1

    return count

print(count_occurances([1, 2, 3, 2, 4, 2, 5], 2))
print(count_occurances([5,5,2,3], 5))
print(count_occurances([1,2,3], 9))
print(count_occurances([], 5) )
```

```python
"""
Find the index of the element
"""

def find_index(nums, target):

    for i , num in enumerate(nums):

        if num == target:
            return i

print(find_index([10, 20, 30, 40, 50], 30))
```

```python
"""
find index of the number using range
"""

def find_index(nums, target):

    for i in range(len(nums)):

        if nums[i] == target:
            return i

    return -1

print(find_index([10, 20, 30, 40, 50], 30))
print(find_index([10, 20, 30, 40, 50], 70))
```

```python
"""
Count Positive, Negative and Zero
"""

def count_numbers(nums):

    positive = 0
    negative = 0
    zero = 0

    for num in nums:

        if num == 0:
            zero += 1
        elif num < 0:
            negative += 1
        else:
            positive += 1

    return positive, negative, zero

print(count_numbers([10, -5, 0, 7, -2, 0, 8, -1]))
```

```python
"""
Sum of All Elements
"""

def sum_numbers(nums):

    total = 0

    for num in nums:
        total += num

    return total

print(sum_numbers([10, 20, 30, 40]))
print(sum_numbers([-5, 10, -2, 7]))
```

```python
"""
Find the Second Largest Element
"""
def second_largest(nums):

    largest = nums[0]
    second = nums[0]

    for num in nums:

        if num > largest:
            
            second = largest
            largest = num

        elif num > second  and num != largest:
            second = num

    return second


print(second_largest([10, 5, 27, 3, 19]))
```

```python
"""
Problem 8 — Reverse a List
"""

def reverse_list(nums):

    left = 0
    right = len(nums) - 1

    while left < right:

        nums[left], nums[right] = nums[right], nums[left]
        left += 1
        right -= 1

    return nums

print(reverse_list([1, 2, 3, 4, 5]))
```

```python
"""
Check if a List is a Palindrome 
"""

def is_palindrome(nums):

    left = 0
    right = len(nums) - 1

    while left < right:

        if nums[left] != nums[right]:
           return False

        left += 1
        right -= 1

    return True

print(is_palindrome([1, 2, 3, 2, 1]))
print(is_palindrome([1, 2, 3, 4]))
print(is_palindrome([5,5,5]))
print(is_palindrome([7]))
print(is_palindrome([]))
```

```python
"""
Duplicate Detection
"""
count = 0

def has_duplicate(nums):

    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] == nums[j]:
                return True

    return False

print(has_duplicate([1, 2, 3, 2, 4, 5]))
print(has_duplicate([1, 2, 3, 4, 5]))
print(has_duplicate([7]))
print(has_duplicate([]))
```

```python
def remove_duplicates(nums):

    unique = []

    for num in nums:
        if num not in unique:
            unique.append(num)
    return unique

print(remove_duplicates([1, 2, 2, 3, 4, 4, 5]))
```

```python
"""
Find the largest difference
"""

def largest_difference(nums):

    maximum = nums[0]
    minimum = nums[0]

    for num in nums:

        if num > maximum:
            maximum = num

        if num < minimum:
            minimum = num

    return maximum - minimum


print(largest_difference([10, 3, 7, 2, 15, 6]))
```

```python
"""
Move all zeroes to the end
"""
def move_zeros(nums):

    left = 0

    for right in range(len(nums)):

        if nums[right] != 0:

            nums[left] ,nums[right] = nums[right], nums[left]

            left += 1
           

    return nums

print(move_zeros([0, 1, 0, 3, 12]))
```

```python
"""
Find missing number
"""

def find_missing(nums):

    for num in range(0,len(nums)):

        if num not in nums:
            nums.append(num)
    return sorted(nums)

print(find_missing([3, 0, 1]))
print(find_missing([9, 6, 4, 2, 3, 5, 7, 0, 1]))
print(find_missing([0]))
```

```python
"""
Find missing number (2nd METHOD)
"""

def find_missing(nums):

    n = len(nums)
    actual = 0
    expected = n * (n + 1) // 2

    for num in nums:
        
        actual += num

    return expected - actual

print(find_missing([9, 6, 4, 2, 3, 5, 7, 0, 1]))
print(find_missing([0]))
print(find_missing([3, 0, 1]))
```

```python
"""
Move Negative Numbers to the Left
"""

def move_negatives(nums):

    left = 0

    for right in range(len(nums)):

        if nums[right] < 0:

            nums[left] , nums[right] = nums[right], nums[left]

            left += 1

    return nums

print(move_negatives([2, -3, 5, -1, 4, -6]))
```

```python
"""
Matrix manipulation
"""

def sum_matrix(matrix):

    total = 0
    
    for row in matrix:
        for num in row:
            total += num

    return total

print(sum_matrix([
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]))
```

```python
"""
count even number in matrix
"""

def count_even_matrix(matrix):

    count = 0

    for row in matrix:
        for num in row:
            if num % 2 == 0:
                count += 1

    return count

print(count_even_matrix([
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]))
```

```python
"""
find maximum in matrix
"""
def max_matrix(matrix):

    max = matrix[0][0]

    for row in matrix:
        for num in row:
            if num > max:
                max = num

    return max

print(max_matrix([
    [10, 25, 3],
    [44, 5, 18],
    [7, 90, 12]
]))
```
