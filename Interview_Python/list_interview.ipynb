```python
"""
max,min number 
"""

def max_num(nums: list):

    max = nums[0]

    for num in nums:
        if num > max:
            max = num

    return max

print(max_num([-5, -2, -10, -1]))
print(max_num([10, 5, 27, 3, 19]))
print(max_num([7]))
```

```python
"""
Min Value
"""

def min_nums(nums:list) -> list:


    if not nums:
        return 1

    min = nums[0]

    for num in nums:
        if num < min:
            min = num

    return min

print(min_nums([10, 5, 27, 3, 19]))
print(min_nums([-5, -2, -10, -1]))
```
