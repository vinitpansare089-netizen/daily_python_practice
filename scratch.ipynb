```python
"""
TWO POINTERS Self Practice
"""


def reverse_string(s: str) -> str:

    s = list(s)

    left = 0
    right = len(s) - 1

    while left < right:

            s[left], s[right] = s[right], s[left]

            left += 1
            right -= 1
            
    return ''.join(s)


name = reverse_string('vinit pansare')
print(name)
```
