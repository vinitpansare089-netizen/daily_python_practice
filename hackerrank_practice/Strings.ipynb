```python
"""
SwapCase program
"""


def to_case(s):
    result = ''

    for char in s:

        if char.islower():
            result += char.upper()
        else:
            result += char.lower()

    return result

v = "vInEEt"
result = to_case(v)
print(result)
```

```python
"""
Split and join a string
"""

def split_and_join(line):

    line = line.split(" ")
    line = '-'.join(line)

    return line

l = input()
print(split_and_join(l))
```

```python
def print_full_name(first, last):
    
    return f'Hello {first} {last}! You just delved into python.'

name = input()
last = input()

print(print_full_name(name, last))
```

```python
def mutate_string(string, position, character):
    string = list(string)
    string[position] = character
    string = ''.join(string) 
    return string

s = input()
l = int(input())
c = input()

print(mutate_string(s,l,c))
```
