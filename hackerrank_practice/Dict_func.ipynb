```python
from collections import defaultdict

def group_numbers(nums):
    d = defaultdict(list)

    for num in nums:
        key = 'div_by_3' if num % 3 == 0 else 'not_div_by_3'
        d[key].append(num)

    return dict(d)
    
nums = [12, 25, 30, 7, 18]
print(group_numbers(nums))
```

```python
from collections import defaultdict

def group_numbers(nums, check):
    d = defaultdict(list)

    for num in nums:
        key = check(num)
        d[key].append(num)

    return dict(d)

def check(num):
    return 'divisible_by_3' if num % 3 == 0 else 'not_divisible_by_3'

def size_check(nums):
    return 'small' if nums <= 15 else 'large'

nums = [12, 25, 30 ,7, 18]
print(group_numbers(nums, size_check))
```
