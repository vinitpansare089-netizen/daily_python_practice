```python
if __name__ == '__main__':
    
    students = []
    
    for students in range(int(input())):
        name = input()
        score = float(input())
        students.append([name, score])


#access the [1] index in list
grades = [student[1] for student in students]

#sorted grades avoid similar list only unique
unique_grades = sorted(set(grades))

#in [1] index..,access element and store it in a variable 
second_unique = unique_grades[1]

#now names store value if 
names = [student[0] for student in students if names == second_unique]

names.sort()

for name in names:
    
    print(name)
```
