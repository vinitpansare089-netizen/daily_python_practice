```python
from collections import defaultdict

nums = [1,2,3,4,5]

d = defaultdict(list)

for num in nums:
    if num % 2 == 0:
        d['Even'].append(num)
    else:
        d['Odd'].append(num)

print(d)



```

```python
from collections import defaultdict

d = defaultdict(int)

for word in d:
    d[word] += 1

print(d)
```

```python
nums = [1,2,3,4,5,6]

d = defaultdict(list)

for num in nums:
    if num % 2 == 0:
        d['even'].append(num)
    else:
        d['odd'].append(num)

print(dict(d))
```

```python
from collections import defaultdict

nums = [1,2,3,4,5,6]

d = defaultdict(list)

for num in nums:
    if num <= 3:
        d['small'].append(num)
    else:
        d['large'].append(num)

print(dict(d))
```

```python
from collections import defaultdict

nums = [1,2,3,4,5,6]

d = defaultdict(list)

for num in nums:
    key = "small" if num <= 3 else "large"
    d[key].append(num)

print(dict(d))
```
